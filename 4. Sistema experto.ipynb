```python
# ------------------------------------------------------------
# 2. OBTENER LOS HECHOS DEL PROBLEMA
# ------------------------------------------------------------

def obtener_hechos():

    print("\n======================================")
    print("============INTERFAZ==================")
    print("  SISTEMA EXPERTO: DIAGNÓSTICO DE PC Y RED")
    print("======================================")

    hechos = set()

    enciende = input("¿La computadora enciende normalmente? (s/n): ").lower()
    if enciende == "n":
        hechos.add("no_enciende")

    corriente = input("¿El tomacorriente o regulador tiene corriente? (s/n): ").lower()
    if corriente == "n":
        hechos.add("sin_corriente")

    olor = input("¿Se percibe olor a quemado cerca del equipo? (s/n): ").lower()
    if olor == "s":
        hechos.add("olor_quemado")

    negro = input("¿El monitor queda en negro al encender? (s/n): ").lower()
    if negro == "s":
        hechos.add("pantalla_negra")

    bips = input("¿Se escuchan pitidos (beeps) al encender? (s/n): ").lower()
    if bips == "s":
        hechos.add("sonido_bips")

    monitor = input("¿El monitor externo está encendido (LED)? (s/n): ").lower()
    if monitor == "s":
        hechos.add("monitor_corriente")

    azul = input("¿Aparece la pantalla azul de error (BSOD)? (s/n): ").lower()
    if azul == "s":
        hechos.add("pantalla_azul")

    driver = input("¿Instalaste un driver o programa justo antes del problema? (s/n): ").lower()
    if driver == "s":
        hechos.add("tras_instalar")

    apaga = input("¿El equipo se apaga solo sin razón aparente? (s/n): ").lower()
    if apaga == "s":
        hechos.add("se_apaga_sola")

    caliente = input("¿El equipo se siente muy caliente al tacto? (s/n): ").lower()
    if caliente == "s":
        hechos.add("equipo_caliente")

    lento = input("¿El equipo está anormalmente lento? (s/n): ").lower()
    if lento == "s":
        hechos.add("equipo_lento")

    disco = input("¿El disco duro tiene menos de 5 GB libres? (s/n): ").lower()
    if disco == "s":
        hechos.add("disco_lleno")

    ventanas = input("¿Se abren ventanas o programas sin que tú hagas nada? (s/n): ").lower()
    if ventanas == "s":
        hechos.add("ventanas_raras")

    internet = input("¿Tu computadora tiene internet? (s/n): ").lower()
    if internet == "n":
        hechos.add("sin_internet")

    router = input("¿El router enciende sus luces? (s/n): ").lower()
    if router == "n":
        hechos.add("router_sin_luces")

    otro = input("¿Otro dispositivo (celular) sí navega con el mismo router? (s/n): ").lower()
    if otro == "s":
        hechos.add("otro_si_navega")

    wifi = input("¿El Wi-Fi de la PC aparece activo? (s/n): ").lower()
    if wifi == "n":
        hechos.add("wifi_apagado")

    cable = input("¿El cable de red está flojo o deteriorado? (s/n): ").lower()
    if cable == "s":
        hechos.add("cable_flojo")

    audio = input("¿Hay sonido en los parlantes o audífonos? (s/n): ").lower()
    if audio == "n":
        hechos.add("sin_audio")

    return hechos
```

```python
# ------------------------------------------------------------
# 3. MOTOR DE INFERENCIA
# ------------------------------------------------------------

def motor_inferencia(hechos):
    """
    Compara los hechos conocidos con las condiciones
    de cada regla.

    Si todas las condiciones de una regla están presentes,
    la regla se activa y genera un diagnóstico.
    """

    resultados = []

    print("\n======================================")
    print("        MOTOR DE INFERENCIA")
    print("======================================")

    print("\nHechos detectados:")

    for hecho in hechos:
        print("  •", hecho)

    print("\nAnalizando reglas...\n")

    for regla in reglas:

        # ----------------------------------------------------
        # Verificamos si TODAS las condiciones de la regla
        # están presentes en los hechos.
        # ----------------------------------------------------

        if all(condicion in hechos
               for condicion in regla["condiciones"]):

            print("✓ Regla activada:",
                  regla["nombre"])

            resultados.append(regla)

    return resultados
```

```python
# ------------------------------------------------------------
# 4. MOSTRAR CONCLUSIONES
# ------------------------------------------------------------

def mostrar_resultados(resultados):

    print("\n======================================")
    print("          RESULTADO DEL SISTEMA")
    print("======================================")

    if not resultados:

        print("\nNo se encontró una falla específica.")
        print("Se recomienda realizar una revisión técnica más detallada.")

        return

    for resultado in resultados:

        print("\nDiagnóstico:")
        print("→", resultado["conclusion"])

        print("\nRecomendación:")
        print("→", resultado["recomendacion"])
```

```python
# ------------------------------------------------------------
# 5. PROGRAMA PRINCIPAL
# ------------------------------------------------------------

def main():

    # Obtener información de la computadora
    hechos = obtener_hechos()

    # Ejecutar el razonamiento
    resultados = motor_inferencia(hechos)

    # Mostrar las conclusiones
    mostrar_resultados(resultados)


main()
```
